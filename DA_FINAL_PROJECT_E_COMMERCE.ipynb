Dataset Description:

Order_id - Unique identification number assigned to each order.

Order_date - The date when the customer placed the order.

Country - Country where the customer is located.

State - State or region of the customer.

City - City where the customer lives.

Product_id - Unique identification number for each product.

Product_name - Name of the product purchased.

Category - Main category to which the product belongs (e.g., Electronics, Clothing,Fashion).

Sub_category - Sub-division of the main product category.

Brand - Brand or manufacturer of the product.

Quantity - Number of units of the product purchased in the order.

Unit_price - Price of a single unit of the product.

Discount - Discount applied to the product price.

Shipping_cost - Cost charged for delivering the product.

Total_sales - Total sales amount for the order after including quantity, discount, and shipping cost.

Payment_method - Method used by the customer to make payment (Card,COD, UPI,NetBanking).

Prescriptive Analysis - Prescriptive analysis provides recommendations based on the analysis results.It suggests strategies such as optimizing discounts, improving product focus, and targeting customers.This helps businesses improve profitability and overall sales performance.

Handle missing values (impute or drop)

```python
#summary statistics
df.iloc[:,17:].describe()
```

```python
#Outliers
sns.boxplot(x=df['total_sales'])
plt.show()
```

```python
#Calculate skewness of all numeric columns
df.skew(numeric_only=True)
```

```python
#Log transformation
df['total_sales_log'] = np.log1p(df['total_sales'])
```

```python
#Check skewness
df[['total_sales', 'total_sales_log']].skew()
```

```python
#convert data types

df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
```

```python
#delivery days
df['delivery_days']=(df['delivery_date']-df['order_date']).dt.days
```

```python
df[['order_date','delivery_date','delivery_days']]
```

```python
#derived features-revenue after discount
df['final_price']=df['unit_price']-df['discount']
```

```python
#view specific columns
df[['unit_price','discount','final_price']].head()
```

```python
#total revenue
df['calculated_sales']=df['quantity']*df['final_price']
```

```python
#New features data converstion for unit_price,discount,total_sales
df['total_sales_round']=df['total_sales'].round(2)
df['unit_price_round']=df['unit_price'].round(2)
df['discount_percent']=(df['discount']*100).round(0)
```

```python
df.head()
```

```python
#column names
df.columns

```

```python
df.to_csv('cleaned_amazon_dataset.csv',index=False)
```
